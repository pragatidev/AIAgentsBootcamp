## 🔐 Step 1: Load OpenAI API Key

```
from pathlib import Path
import sys

def _find_root():
    hints = [
        Path("data") / "udemy" / "courses" / "ai_agents_bootcamp" / "course_repo",
        Path("course_repo"),
    ]
    def ok(p: Path) -> bool:
        return (p / "src" / "llm.py").is_file() or (p / "Section_5_Autonomous_Workflows").is_dir()
    cur = Path.cwd().resolve()
    for p in [cur, *cur.parents]:
        if ok(p):
            return p
        for h in hints:
            cand = p / h
            if ok(cand):
                return cand.resolve()
    raise FileNotFoundError("Open this notebook from the AIAgentsBootcamp folder, or set cwd to that repo.")

_ROOT = _find_root()
if str(_ROOT) not in sys.path:
    sys.path.insert(0, str(_ROOT))
from src.llm import REPO_ROOT, get_agent, get_autogen_config, get_embeddings, get_llm
llm = get_llm()
```

## 📦 Step 2: Import PromptTemplate and OpenAI LLM

```
from langchain_core.prompts import PromptTemplate
```

## ✏️ Step 3: Design a specific prompt with constraints

```
prompt = PromptTemplate(
    input_variables=["topic"],
    template="Summarize {topic} in 20 words or less."
)
```

## 🔗 Step 4: Create the LangChain pipeline

```
llm = get_llm()
chain = prompt | llm
```

## 🚀 Step 5: Try with a topic

```
response = chain.invoke({"topic": "Machine Learning"})
print("Response:", response)
```
